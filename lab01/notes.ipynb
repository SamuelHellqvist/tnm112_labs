Train loss:     0.0095
	Train accuracy: 99.02
	Test loss:      0.0094
	Test accuracy:  99.02

The math behind ReLU is more suited for this classification problem. We dont get negative values, we get the max(0, x). In the sigmoid function we can have negative values and in the linear we get only x. 

In sigmoid the answer is always between 0 and 1. This makes the method better for linear classifications. Sigmoid therefore have a problem when we have hidden layers since large negatuve values produce gradients close to 0. ReLU does not have this problem and therefore perform better when we have hidden layers. 

Note: we might see some overfitting at the ReLU implementation.
